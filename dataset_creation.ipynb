```python
import pandas as pd
from cleaning import remove_outliers, remove_duplicates, handle_missing, create_cleaning_report
```

```python
df = pd.read_csv("learning_traces.13m.csv")
df.head()
```

```python
df_before = df.copy()

df_unique = remove_duplicates(df)
df_no_na = handle_missing(df_unique, "rows")
df_norm = remove_outliers(df_no_na, ["p_recall", "history_seen", "history_correct", "session_seen", "session_correct"], "iqr", 1.5)

df_final = df_norm.copy()
print(create_cleaning_report(df_before, df_final))
```

```python
df_sampled = df.sample(n=5000)
df_sampled.to_csv("analysis_dataset.csv", index=False)
```
